```python
import torch
import json
import matplotlib.pyplot as plt
import pandas as pd
import os
import re
from src import generation, model, datasets, tokenizer as token
```

```python
MODEL_DIR = './runs'
MODEL_RUN = 'base'
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
def load_assets(model_dir, model_run=None):
    if not model_run:
        dirs = os.listdir(model_dir)
        model_runs = [d for d in dirs if re.fullmatch(r"^\d{8}_\d{6}$", d)]
        model_runs.sort()
        model_run = model_runs[-1]
    with open(f'{model_dir}/{model_run}/results.json', 'r') as f:
        results = json.load(f)
        results_df = pd.DataFrame(results)
    with open(f'{model_dir}/{model_run}/config.json', 'r') as f:
        config = json.load(f)

    return results_df, config
```

```python
results_df, config = load_assets(MODEL_DIR, MODEL_RUN)
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

train = results_df[results_df["type"] == "train"]
val = results_df[results_df["type"] == "validation"]

ax.plot(train["step_no"], train["loss"], alpha=0.3, label="Train")
ax.plot(val["step_no"], val["loss"], marker="o", label="Validation")
ax.set_xlabel("Step")
ax.set_ylabel("Loss")
ax.set_title("Training vs Validation Loss")
ax.legend()
ax.set_xscale("log")
ax.set_yscale("log")
plt.tight_layout()
plt.show()
```
